```python
important_columns = [
    "Order Id",
    "order date (DateOrders)",
    "shipping date (DateOrders)",
    "Delivery Status",
    "Late_delivery_risk",
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Shipping Mode",
    "Customer Id",
    "Customer City",
    "Customer State",
    "Customer Country",
    "Customer Segment",
    "Product Card Id",
    "Product Name",
    "Category Name",
    "Department Name",
    "Product Price",
    "Order Region",
    "Order Country",
    "Order State",
    "Order City",
    "Market",
    "Sales",
    "Order Profit Per Order",
    "Order Item Quantity",
    "Order Item Discount"
]

df = df[important_columns]
```

```python
df.shape
```

```python
df.isnull().sum() #check whether the dataset contains any missing (empty) values.
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col].fillna("Unknown", inplace=True) #used to find all text (string) columns in the dataset and replace any missing (NULL) values with the word "Unknown".
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col].fillna(df[col].median(), inplace=True) #used to find all numeric columns in the dataset and replace any missing (NULL/NaN) values with the median of that column.
```

```python
df.duplicated().sum() #Check how many duplicate rows are present in the dataset.
```

```python
df.drop_duplicates(inplace=True) #It removes duplicate rows from the DataFrame. inplace=true means Update the original DataFrame directly.
```

```python
df.shape
```

```python
df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"]
)


df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"] # use to convert the object or text datatype to a datetime because python can not able to calculate between text
)
```

```python
df.info()
```

```python
df["Delay Days"] = (
    df["Days for shipping (real)"]
    -
    df["Days for shipment (scheduled)"]
)#code creates a new column called "Delay Days".
```

```python
df[[
"Days for shipping (real)",
"Days for shipment (scheduled)",
"Delay Days"
]].head() #checking whether the newly created Delay Days column has been calculated correctly.
```

```python
df["Delay Status"] = np.where(
    df["Delay Days"] > 0,
    "Delayed",
    "On Time"
) #If Delay Days > 0, the order is marked as "Delayed".
#Otherwise, it is marked as "On Time".
```

```python
df["Delay Status"].value_counts() #How many orders were Delayed
#How many orders were On Time
```

```python
def risk_category(delay):

    if delay <= 0:
        return "Low Risk"

    elif delay <= 3:
        return "Medium Risk"

    else:
        return "High Risk"


df["Risk Level"] = df["Delay Days"].apply(risk_category)
```
